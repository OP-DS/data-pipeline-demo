```python
import time
from datetime import datetime
from pathlib import Path

from saldo_functions_OD import (
    load_data,
    clean_station_names,
    add_sequence,
    compute_saldo,
    build_final_table,
    compute_pkm,
    export_data,
)

print("----------------------------------------------")
print(
    "Starting data pipeline at",
    datetime.now().strftime("%Y-%m-%d %H:%M:%S")
)
print("----------------------------------------------")


INPUT_PATH = Path("input") / "ondemand_07_2026.xlsx"
OUTPUT_PATH = Path("data") / "final_07_V2.xlsx"

OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)


# Step 1: Load data
t0 = time.time()

df = load_data(INPUT_PATH)

t1 = time.time()
print("Step 1: Done")
print(f"---> Data read in {t1 - t0:.2f} seconds\n")


# Step 2: Clean station names
t0 = time.time()

df = clean_station_names(df)

t1 = time.time()
print("Step 2: Done")
print(f"---> Station names cleaned in {t1 - t0:.2f} seconds\n")


# Step 3: Add sequence
t0 = time.time()

df = add_sequence(df)

t1 = time.time()
print("Step 3: Done")
print(f"---> SQ added in {t1 - t0:.2f} seconds\n")


# Step 4: Compute saldo
t0 = time.time()

df = compute_saldo(df)

t1 = time.time()
print("Step 4: Done")
print(f"---> Saldo calculated in {t1 - t0:.2f} seconds\n")


# Step 5: Build final table
t0 = time.time()

df_final = build_final_table(df)

t1 = time.time()
print("Step 5: Done")
print(f"---> Final table set in {t1 - t0:.2f} seconds\n")


# Step 6: Compute PKM
t0 = time.time()

df_final = compute_pkm(df_final)

t1 = time.time()
print("Step 6: Done")
print(f"---> PKM calculated in {t1 - t0:.2f} seconds\n")


# Step 7: Export
t0 = time.time()

export_data(df_final, OUTPUT_PATH)

t1 = time.time()
print("Step 7: Done")
print(f"---> File exported in {t1 - t0:.2f} seconds\n")


print("----------------------------------------------")
print(
    "Pipeline finished at",
    datetime.now().strftime("%Y-%m-%d %H:%M:%S")
)
```
